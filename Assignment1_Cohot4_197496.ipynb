```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data ['Glucose'] = le.fit_transform(data['Glucose'])
data ['Insulin'] = le.fit_transform(data['Insulin'])
data ['BMI'] = le.fit_transform(data['BMI'])
data ['DiabetesPedigreeFunction'] = le.fit_transform(data['DiabetesPedigreeFunction'])
data ['Age'] = le.fit_transform(data['Age'])
data ['Outcome'] = le.fit_transform(data['Outcome'])
```

```python
x=data.drop(['Outcome'],axis=1) #seperating dataset into features and target variable
y=data['Outcome']
```

```python
#Train test-split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
#Independant and Dependant columns have been identified based on the column details in the file which helps to determine diabetes patients
#and the logistic Regression model is trained using both the training and testing dataset
```

```python
model.fit(x_train,y_train)  #train the logistic regression model
```

```python
predictions_test=model.predict(x_test)  #generate predictions on the training dataset
predictions_train=model.predict(x_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score=accuracy_score(y_test,predictions_test) #evaluate model's accuracy
```

```python
print(accuracy_score)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,predictions_test))
```

```python
from sklearn.metrics import confusion_matrix
print("Confusion matrix:",confusion_matrix(y_test,predictions_test))
```

```python
from sklearn.metrics import accuracy_score
#Predictions
train_accuracy = accuracy_score(y_train,predictions_train)
test_accuracy = accuracy_score(y_test,predictions_test)
```

```python
print("Train accuracy:",train_accuracy)
print("Test accuracy:",test_accuracy)
```

```python
#The training accuracy percentage is 76, which shows here how well the model used the data to train.
#The model has a testing accuracy of 76% as well.The difference is minimal indicating the model performs consistently
#The Confusion matrix implies that it has identified 95 negatives
```
